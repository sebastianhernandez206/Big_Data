```python
resultado = (
    df.groupby("pais")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
resultado.head(10)
```

```python
year_df = df[(df["year"] >= 1970) & (df["year"]<= 2023)]
print(year_df)
```

```python
resultado = (
    year_df.groupby("pais")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)

print(resultado)
```
