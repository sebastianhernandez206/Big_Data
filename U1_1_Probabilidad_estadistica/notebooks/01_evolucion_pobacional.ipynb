```python
# Filtrar los datos para el año histórico 1951
df_1951 = dataframe[dataframe["Year"] == 1951].dropna(subset=["Annual change in population"])

# Variable sobre la que aplicaremos el análisis[cite: 2]
poblacion_1951 = df_1951["Annual change in population"]

# Cálculos estadísticos[cite: 2]
media = poblacion_1951.mean()
mediana = poblacion_1951.median()
varianza = poblacion_1951.var()
desviacion_std = poblacion_1951.std()
minimo = poblacion_1951.min()
maximo = poblacion_1951.max()

# Identificar entidades del mínimo y máximo[cite: 2]
entidad_min = df_1951.loc[poblacion_1951.idxmin()]["Entity"]
entidad_max = df_1951.loc[poblacion_1951.idxmax()]["Entity"]

# Imprimir resultados
print(f"Año seleccionado: 1951")
print(f"Media: {media:,.2f}")
print(f"Mediana: {mediana:,.2f}")
print(f"Varianza: {varianza:,.2f}")
print(f"Desviación Estándar: {desviacion_std:,.2f}")
print(f"Mínimo: {minimo:,.0f} ({entidad_min})")
print(f"Máximo: {maximo:,.0f} ({entidad_max})")
```
